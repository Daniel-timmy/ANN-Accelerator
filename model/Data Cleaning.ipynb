```python
import pandas as pd
import numpy as np

import seaborn as sns
import matplotlib.pylab as plt
%matplotlib inline 

from scipy import stats
import joblib
```

```python
data = pd.read_csv('ourDataSheet.csv')
data.head()
```

```python
data.dropna(inplace = True)
```

```python
data.describe()
```

```python
data.info()
```

```python
df = data.select_dtypes(include =['int64','float64'])
df
```

```python
data_list = df.values.tolist()
data_cleaned = []
for i in data_list:
    if i[-1] == 1.0:
        if i[0] >= 40.0 or i[6] >= 40.0 or i[12] >= 40.0 or i[18] >= 40.0:
            if i[1] <= 60 or i[7] <= 60  or i[13] <= 60  or i[19] <= 60:
                data_cleaned.append(i)
    else:
        if i[0] < 40.0 and i[6] < 40.0 and i[12] < 40.0 and i[18] < 40.0:
            if i[1] > 60 and i[7] > 60  and i[13] > 60  and i[19] > 60:
                data_cleaned.append(i)
                
print(len(data_cleaned))
cleaned_data = pd.DataFrame(data_cleaned, columns=df.columns)
                
```

```python
model_data = cleaned_data.drop(columns=["L1", "L2", "L3", "L4"])
```

```python
model_data.info()
```

```python
model_data.to_csv('model_data.csv',index = False)
```

```python
from random import shuffle

with open('model_data.csv') as ip:
    lines=ip.readlines()
    header = lines.pop(0)
    shuffle(lines)
    lines.insert(0, header)

with open('shuffled_model_data.csv','w') as out:
    out.writelines(lines)
```

```python
stats_data = pd.read_csv("shuffled_model_data.csv")
numeric_columns = stats_data.select_dtypes(include="number").columns.drop("CLASS_LABEL")

class_statistics = (
    stats_data.groupby("CLASS_LABEL")[numeric_columns]
    .agg(["min", "max", "mean"])
    .reindex([0, 1])
)

class_statistics
```
